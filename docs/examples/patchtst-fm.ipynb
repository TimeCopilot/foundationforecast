# Using PatchTST-FM

Forecast with PatchTST-FM using `foundationforecast`.

Requires Python 3.11–3.13.


```python
import pandas as pd
from foundationforecast import FoundationForecast
from foundationforecast.models import PatchTSTFM
```

## Load data


```python
df = pd.read_csv(
    "https://timecopilot.s3.amazonaws.com/public/data/air_passengers.csv",
    parse_dates=["ds"],
)
df["unique_id"] = "AirPassengers"
```

## Forecast with PatchTST-FM


```python
ff = FoundationForecast(models=[PatchTSTFM()])
fcst_df = ff.forecast(df=df, h=12, freq="MS")
fcst_df.head()
```

```python
from foundationforecast.core.forecaster import Forecaster

Forecaster.plot(df, fcst_df)
```
