foundationforecast ships a `Tabby` class for [Huawei Paris Noah's Ark Lab](https://huggingface.co/paris-noah)'s [Tabby](https://huggingface.co/paris-noah/Tabby) checkpoint ([arXiv:2609.13956](https://arxiv.org/abs/2609.13956)). The released weights are a self-contained prompted PatchTST-style model (~147M) loaded with `transformers.AutoModel` and `trust_remote_code=True`.

Tabby forecasts each series independently with 99 native quantile knots (0.01–0.99). The backbone window is 8,192 steps; for horizon `H` it masks `max(H, 128)` future positions. It does not accept known-future covariates — use `exog_strategy=False` in mixed `FoundationForecast` runs.

*Requirements:*

    - Python 3.10+
    - `transformers` and `torch` (installed with foundationforecast)

**License:** checkpoint weights are [CC BY-NC 4.0](https://creativecommons.org/licenses/by-nc/4.0/) (non-commercial).


## Import libraries


```python
import pandas as pd

from foundationforecast import FoundationForecast
from foundationforecast.models import Tabby
```

## Load the dataset

The DataFrame must include at least the following columns:
- unique_id: Unique identifier for each time series (string)
- ds: Date column (datetime format)
- y: Target variable for forecasting (float format)


```python
df = pd.read_csv(
    "https://timecopilot.s3.amazonaws.com/public/data/air_passengers.csv",
    parse_dates=["ds"],
)
df.head()
```

## Create a FoundationForecast


```python
ff = FoundationForecast(
    models=[
        Tabby(alias="Tabby"),
    ]
)
```

## Forecast


```python
fcst_df = ff.forecast(df, h=12, freq="MS", level=[90])
fcst_df.head()
```

```python
ff.plot(df=df, forecasts_df=fcst_df)
```
